```python
import pandas as pd
import numpy as np
from datetime import datetime

# Load raw NPCI data
raw_data = pd.read_csv('../data/raw/npci_upi_data.csv')
print(raw_data.head())
print(raw_data.dtypes)
```

```python
# Cleaning steps:
# 1. Parse dates
raw_data['date'] = pd.to_datetime(raw_data['date'], format='%d-%m-%Y')

# 2. Fix bank names (standardize)
raw_data['bank_name'] = raw_data['bank_name'].str.strip().str.upper()

# 3. Convert to numeric
raw_data['txn_count'] = pd.to_numeric(raw_data['txn_count'], errors='coerce')

# 4. Handle nulls
raw_data = raw_data.dropna(subset=['date', 'bank_name'])

print(f"Cleaned rows: {len(raw_data)}")
raw_data.to_csv('../data/processed/npci_upi_clean.csv', index=False)
```
